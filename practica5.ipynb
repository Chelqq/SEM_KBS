```python
import numpy as np
```

```python
def sigmoid(z):
    """
    un valor entre 0 y 1 que representa la probabilidad de que una observación pertenezca a una cierta clase
    """
    return 1 / (1 + np.exp(-z))

def log_likelihood(X, y, beta):
    """
    mide qué tan bien se ajusta el modelo a los datos. Una mayor verosimilitud indica un mejor ajuste.
    """
    h = sigmoid(X @ beta)
    return np.sum(y * np.log(h) + (1 - y) * np.log(1 - h))

def gradient(X, y, beta):
    """
     Calcula el gradiente de la función de verosimilitud logarítmica con respecto a los parámetros beta
    """
    h = sigmoid(X @ beta)
    return X.T @ (h - y)
```

```python

# Datos de ejemplo
X = np.array([[1, 2], [3, 4], [5, 6]])
y = np.array([0, 1, 1])

# Inicializar parámetros
beta = np.random.randn(X.shape[1])

# Tasa de aprendizaje y número máximo de iteraciones
learning_rate = 0.01
max_iter = 1000

# Gradiente descendente
for i in range(max_iter):
    grad = gradient(X, y, beta)
    beta -= learning_rate * grad

# Predicciones
predictions = sigmoid(X @ beta)
```
